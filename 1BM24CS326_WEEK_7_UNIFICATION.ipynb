```python

def occurs_check(var, term, subst):
    term = substitute(term, subst)

    if var == term:
        return True

    if isinstance(term, tuple):
        return any(occurs_check(var, arg, subst)
                   for arg in term[1:])

    return False


def substitute(term, subst):
    if isinstance(term, str):
        while term in subst:
            term = subst[term]
        return term

    if isinstance(term, tuple):
        return tuple(substitute(arg, subst) for arg in term)

    return term


def unify(x, y, subst=None):
    if subst is None:
        subst = {}

    x = substitute(x, subst)
    y = substitute(y, subst)

    if x == y:
        return subst

    # Check if x is a variable
    if isinstance(x, str) and x.islower():
        if occurs_check(x, y, subst):
            return None
        subst[x] = y
        return subst

    # Check if y is a variable
    if isinstance(y, str) and y.islower():
        if occurs_check(y, x, subst):
            return None
        subst[y] = x
        return subst

    # Check predicates and arguments
    if isinstance(x, tuple) and isinstance(y, tuple):
        if x[0] != y[0] or len(x) != len(y):
            return None

        for a, b in zip(x[1:], y[1:]):
            subst = unify(a, b, subst)

            if subst is None:
                return None

        return subst

    return None


# Convert user input into predicate format
def get_expression():
    predicate = input("Enter predicate name: ").strip()
    arguments = input("Enter arguments separated by spaces: ").split()

    return tuple([predicate] + arguments)


# Main program
print("UNIFICATION ALGORITHM")

print("\nEnter First Expression:")
expr1 = get_expression()

print("\nEnter Second Expression:")
expr2 = get_expression()

result = unify(expr1, expr2)

if result is not None:
    print("\nUnification Successful!")
    print("Substitution:", result)

    print("Unified Expression:",
          substitute(expr1, result))
else:
    print("\nUnification Failed!")
```
