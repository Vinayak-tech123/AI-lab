```python
from itertools import product

# Function to check logical implication
def implies(a, b):
    return (not a) or b

# Knowledge Base
def check_kb(P, Q, R):
    s1 = implies(Q, P)       # Q -> P
    s2 = implies(P, not Q)   # P -> NOT Q
    s3 = Q or R              # Q OR R

    return s1, s2, s3, s1 and s2 and s3


# Generate all possible truth assignments
print("P Q R | Q->P | P->~Q | Q|R | KB")
print("-" * 38)

models = []

for P, Q, R in product([True, False], repeat=3):

    s1, s2, s3, kb = check_kb(P, Q, R)

    # Print truth table
    values = [P, Q, R, s1, s2, s3, kb]
    print(" ".join("T" if x else "F" for x in values))

    # Store models where KB is true
    if kb:
        models.append((P, Q, R))


# Check whether KB entails a query
def entails(query):
    for P, Q, R in models:
        if not query(P, Q, R):
            return False
    return True


# Queries
print("\nModels where KB is TRUE:", models)

print("ii) KB entails R:", entails(lambda P, Q, R: R))

print("iii) KB entails R -> P:",
      entails(lambda P, Q, R: implies(R, P)))

print("iv) KB entails Q -> R:",
      entails(lambda P, Q, R: implies(Q, R)))
```
