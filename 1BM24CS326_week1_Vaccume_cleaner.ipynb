Vacuum Cleaner - Simple Reflex Agent


```python
# Vacuum Cleaner - Simple Reflex Agent

room = {
    "A": "Dirty",
    "B": "Dirty"
}

location = "A"


def simple_reflex_agent(location, room):

    if room[location] == "Dirty":
        return "Suck"

    elif location == "A":
        return "Move Right"

    else:
        return "Move Left"


while True:

    print("\nLocation:", location)
    print("Room A:", room["A"])
    print("Room B:", room["B"])

    action = simple_reflex_agent(location, room)

    print("Action:", action)

    if action == "Suck":
        room[location] = "Clean"

    elif action == "Move Right":
        location = "B"

    elif action == "Move Left":
        location = "A"

    if room["A"] == "Clean" and room["B"] == "Clean":
        print("All rooms are clean!")
        break
```

Vacuum Cleaner - Goal Based Agent


```python
room = {
    "A": "Dirty",
    "B": "Dirty"
}

location = "A"


def goal_based_agent(location, room):

    # Goal: Both rooms should be clean

    if room["A"] == "Clean" and room["B"] == "Clean":
        return "Stop"

    if room[location] == "Dirty":
        return "Suck"

    elif location == "A" and room["B"] == "Dirty":
        return "Move Right"

    elif location == "B" and room["A"] == "Dirty":
        return "Move Left"

    else:
        return "Stop"


while True:

    print("\nLocation:", location)
    print("Room A:", room["A"])
    print("Room B:", room["B"])

    action = goal_based_agent(location, room)

    print("Action:", action)

    if action == "Suck":
        room[location] = "Clean"

    elif action == "Move Right":
        location = "B"

    elif action == "Move Left":
        location = "A"

    elif action == "Stop":
        print("Goal achieved! All rooms are clean!")
        break
```
